## Topic Description
Excerpt from the latest United Nations World Drug Report (2025) focusing on european drug prices, purities and seizures during 2019-2023. Analyzing trends, and figuring out where to best sell each drug, in order to give recommendations to law enforcement where to increase and focus their seizures.

## Dataset Description
both datasets come from the united nations: https://www.unodc.org/unodc/en/data-and-analysis/world-drug-report-2025-annex.html 

The data is free to use, terms and conditions of use can be found here: https://data.un.org/Host.aspx?Content=UNdataUse 
Data was initially downloaded on 04.11.2025 11:19

### Seizures of illicit drugs (2019-2023)
https://www.unodc.org/documents/data-and-analysis/WDR_2025/Annex/7.1_Drug_seizures_2019-2023.xlsx 

Features: 
- Region: String describing a world region eg. Europe
- SubRegion: String describing a sub region eg. Western and Central European
- Country: String describing a countrys name 
- DrugGroup:  String describing a illicit substance group
- DrugSubGroup: String describing further specifiying the illicit substance
- DrugName: String describing the illicit substance name
- Reference year: Number describing the year this drug seizure for a specific substance was recorded
- Kilograms: Number describing the amount of drugs seizured
- msCode: String describing a abb. for a country eg. AUT

Sample Size: ~ 11k


### Prices and purity of illicit drugs (2019-2023)
https://www.unodc.org/documents/data-and-analysis/WDR_2025/Annex/8.1_Prices_and_purities_of_drugs.xlsx

Prices Features: 
- Region: String describing a world region eg. Europe
- SubRegion: String describing a sub region eg. Western and Central European
- Country/Territory: String describing a countrys name 
- DrugGroup:  String describing a illicit substance group
- Drug: String describing the illicit substance name
- Specify: String describing further specifiying the illicit substance
- LevelOfSale: String describing wheter the price for a substance was recorded as a 'Wholesale' or 'Retail' transaction 
- Year: Number describing the year this drug price for a specific substance was recorded
- Typical_USD: Number describing the typical prices collected for the location and substance in that year
- Minimum_USD: Number describing the minimum prices collected for the location and substance in that year
- Maximum_USD: Number describing the maximum prices collected for the location and substance in that year
- Unit: String describing the measuring unit for the the USD prices collected for the location and substance in that year

Prices Sample Size: ~4.6k

Purity Features: 
- Region: String describing a world region eg. Europe
- SubRegion: String describing a sub region eg. Western and Central European
- Country/Territory: String describing a countrys name 
- LevelOfSale: String describing wheter the purity for a substance was recorded as a 'Wholesale' or 'Retail' 
- DrugGroup:  String describing a illicit substance group
- Drug: String describing the illicit substance name
- Specify: String describing further specifiying the illicit substance
- Year: Number describing the year this drug seizure for a specific substance was recorded
- Typical: Number describing the typical amount measured for the location and substance in that year
- Minimum: Number describing the minimum amount measured for the location and substance in that year
- Maximum: Number describing the maximum amount measured for the location and substance in that year
- Measurement: String describing the measuring unit for the the USD prices collected for the location and substance in that year

Purities Sample Size: ~1.7k
